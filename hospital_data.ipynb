```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/drive/MyDrive/Dataset/hospital_data_unclean.csv')
```

```python
df
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
# Name column correction to standard title
df['Name'] = df['Name'].str.title()
df['Name']
```

```python
df['Gender'] = df['Gender'].replace({"M":"MALE", "F":"FEMALE"})
df['Gender']
```

```python
# You can check the data types of all columns in your DataFrame using df.dtypes.
df.dtypes
```

```python
df['Date of Admission'] = df['Date of Admission'].astype(str)
df['Discharge Date'] = df['Discharge Date'].astype(str)
```

```python
df.dtypes
```

```python
df['Date of Admission'] = df['Date of Admission'].str.replace('.', '-', regex=False)
df['Discharge Date'] = df['Discharge Date'].str.replace('.', '-', regex=False)
df['Date of Admission'] = df['Date of Admission'].str.replace('/', '-', regex=False)
df['Discharge Date'] = df['Discharge Date'].str.replace('/', '-', regex=False)
```

```python
df
```

```python
# replace the '#' character with a space in the 'Doctor' column

df['Doctor'] = df['Doctor'].str.replace('#', ' ', regex=False)
df['Doctor'].head()
```

```python
# replace the commas with spaces in the Hospital column for you.

df['Hospital'] = df['Hospital'].str.replace(",","")
df
```

```python
df['Hospital'] = df['Hospital'].fillna('unknown')
df.isna().sum()
```

```python
df.isna().sum()
```

```python
# Here are the unique items in the 'Insurance Provider' column:

df['Insurance Provider'].unique()
```

```python
#  All instances of 'UnitedHealthcare' should now appear as 'United Healthcare'.

df['Insurance Provider'] = df['Insurance Provider'].str.replace('UnitedHealthcare', 'United Healthcare', regex=False)
```

```python
# fill the null values in the 'Name' column with 'unknown'.

df['Name'] = df['Name'].fillna('unknown')
df['Name'].head()
```

```python
df['Age'].dtype
```

```python
# Calculate the mean of the 'Age' column
age_mean = df['Age'].mean()

# Fill null values in 'Age' column with the calculated mean
df['Age'] = df['Age'].fillna(age_mean)

# Display the first few rows of the 'Age' column to show the effect
df['Age'].head()
```

```python
df
```

```python
df['Gender'].value_counts()
```

```python
# Calculate the mode of the 'Gender' column
gender_mode = df['Gender'].mode()[0]

# Fill null values in 'Gender' column with the calculated mode
df['Gender'] = df['Gender'].fillna(gender_mode)

# Display the value counts of the 'Gender' column to show the effect
df['Gender'].value_counts()
```

```python
# Calculate the mode of the 'Blood Type' column
blood_type_mode = df['Blood Type'].mode()[0]

# Fill null values in 'Blood Type' column with the calculated mode
df['Blood Type'] = df['Blood Type'].fillna(blood_type_mode)

# Display the value counts of the 'Blood Type' column to show the effect
df['Blood Type'].value_counts()
```

```python
# Fill null values in 'Doctor' column with the calculated mode
df['Doctor'] = df['Doctor'].fillna("unknown")

# Display the value counts of the 'Doctor' column to show the effect
df['Doctor'].value_counts()
```

```python
df['Medical Condition'].value_counts()
```

```python
df['Medical Condition'] = df['Medical Condition'].fillna('unknown')
df['Medical Condition'].value_counts()
```

```python
df['Insurance Provider'] = df['Insurance Provider'].fillna('unknown')
df['Insurance Provider'].head()
```

```python
df['Billing Amount'] = df['Billing Amount'].astype(str)
df['Billing Amount'] = df['Billing Amount'].str.replace(" USD","")
df['Billing Amount'].head()
```

```python
df
```

```python
df.isna().sum()
```

```python
df['Medication'] = df['Medication'].fillna('unknown')
df['Medication'].value_counts()
```

```python
df['Medication'].value_counts()
```

```python
df['Test Results'].value_counts()
```

```python
df.isna().sum()
```

```python
df['Test Results'] = df['Test Results'].fillna('unknown')
df['Test Results'].value_counts()
```

```python
df['Room Number'] = df['Room Number'].fillna('not assigned')
df['Room Number'].head()
```

```python
df.isna().sum()
```

```python
df['Admission Type'] = df['Admission Type'].fillna('unknown')
df['Admission Type'].value_counts()
```

```python
df.isna().sum()
```

```python
df['Date of Admission'] = df['Date of Admission'].astype(str).str.strip()
df['Discharge Date'] = df['Discharge Date'].astype(str).str.strip()
dates = pd.to_datetime(df['Date of Admission'], errors='coerce')
```

```python
df
```

```python
mask = dates.isna()
df.loc[mask, 'Date of Admission']
```

```python
# First snippet: Basic string cleaning
df['Blood Type'] = df['Blood Type'].astype(str).str.strip().str.upper()

# Second snippet: Using regex to handle trailing '+' or '-' symbols
df['Blood Type'] = df['Blood Type'].str.replace(r"(\+|-)+$", lambda x: x.group(0)[0].upper(), regex=True)
```

```python
df
```

```python
df['Blood Type'].value_counts()
```

```python
import numpy as np

df['Billing Amount'] = df['Billing Amount'].replace('nan', np.nan) # Replace string 'nan' with actual NaN
df['Billing Amount'] = pd.to_numeric(df['Billing Amount'], errors='coerce').round(2)
df['Billing Amount'].head()
```

```python
df.to_csv('Hospital_data_cleaned.csv',index=False)
```
